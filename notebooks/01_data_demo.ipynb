```python
import sys;sys.path.append('../src')

from data import DogSet,denormalize
from torch.utils.data import DataLoader
from torchvision.utils import make_grid
from torchvision.transforms.v2.functional import to_pil_image

dogset = DogSet(data_dir='../data/afhq' ,image_size= 64)
loader = DataLoader( dataset=dogset , batch_size=16 ,  shuffle=True)
batch = next(iter(loader))
assert batch.shape == (16,3,64,64)
print (batch.min() , batch.max())
vis = denormalize(batch)

grid = make_grid(vis,nrow=4)
img = to_pil_image(grid)
img.show()
```
