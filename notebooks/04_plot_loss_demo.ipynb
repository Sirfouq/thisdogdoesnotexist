```python
import json
import matplotlib.pyplot as plt
import os
```

```python
with open('../checkpoints/ddpm_history.json', 'r') as f:
    ddpm_history = json.load(f)

with open('../checkpoints/vae_history.json', 'r') as f:
    vae_history = json.load(f)

ddpm_train_loss = ddpm_history.get('train_loss' , [])
ddpm_val_loss = ddpm_history.get('val_loss', [])
vae_train_loss = vae_history.get('train_loss' , [])
vae_val_loss = vae_history.get('val_loss', [])

```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

ax[0].plot(ddpm_train_loss, label='train', lw = 1.2)
ax[0].plot(ddpm_val_loss, label='val', lw=1.2, alpha= 0.7)
ax[0].set_yscale('log')
ax[0].set_title('DDPM')
ax[0].set_xlabel('epoch') 
ax[0].set_ylabel('MSE loss')
ax[0].legend(); ax[0].grid(alpha=0.3)

ax[1].plot(vae_train_loss, label='train', lw=1.2)
ax[1].plot(vae_val_loss, label='val', lw= 1.2, alpha = 0.7)
ax[1].set_title('VAE')
ax[1].set_ylim(300,1200)
ax[1].set_xlabel('epoch') 
ax[1].set_ylabel('ELBO (nats/image)')
ax[1].legend(); ax[1].grid(alpha=0.3)

os.makedirs('../assets/', exist_ok=True)
plt.savefig('../assets/curves.png', dpi=130, bbox_inches='tight')
```
